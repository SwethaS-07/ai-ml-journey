```python
import numpy as np
```

```python
random = np.random.default_rng(seed=1)
```

```python
print(random.integers(1,31))
```

```python
print(random.integers(low =1,high=31))
```

```python
print(random.integers(low =1,high=31,size = 5))
```

```python
print(random.integers(low =1,high=31, size= (3,2)))
```

```python
print(np.random.uniform(low=-1,high=1))
```
